[leafmap](https://leafmap.org) can put the Cloud-Optimized GeoTIFFs straight on an interactive map. The tiles are rendered by a public TiTiler service, so no data is downloaded to the notebook — and the first view of an area can take a few seconds.

Values are stored in decimetres, so a 0–50 m colour range is `rescale="0,500"`.

```python
%pip install -q leafmap
```

```python
import leafmap

STORE = "https://data.source.coop/geoai-ucph/gvsm"
tile = "32UNA"  # Spessart, Germany

m = leafmap.Map(center=[49.937, 9.312], zoom=12)
m.add_basemap("Esri.WorldImagery")
m.add_cog_layer(
    f"{STORE}/2020/{tile}/RH98_Q1.tif",
    name="RH98 (10 m)",
    colormap_name="inferno",
    rescale="0,500",
    nodata=32767,
)
m.add_cog_layer(
    f"{STORE}/2020/{tile}/RH25_Q1.tif",
    name="RH25 (10 m)",
    colormap_name="inferno",
    rescale="0,120",
    nodata=32767,
    shown=False,
)
m
```

Toggle the layers in the layer control (top right) to compare RH98 (≈ canopy top, 0–50 m) with RH25 (0–12 m). For the whole world at ~1 km, use the overview mosaic:

```python
world = leafmap.Map(center=[10, 0], zoom=2)
world.add_cog_layer(
    f"{STORE}/mosaics/2020/RH98_Q1.tif",
    name="RH98 (~1 km)",
    colormap_name="inferno",
    rescale="0,500",
    nodata=32767,
)
world
```
